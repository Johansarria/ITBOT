## 4. Ingeniería de Características: Creando la Variable Objetivo

El paso más importante. Necesitamos definir qué constituye una "operación exitosa". Para una operación de compra, podemos decir que fue exitosa si el precio subió un cierto porcentaje `X` después de un cierto período de tiempo `Y`.

**Lógica Propuesta:**
1.  Convertir las columnas de `timestamp` a formato datetime.
2.  Para cada fila en `df_analysis` que resultó en una decisión de 'posible compra', miraremos el precio `N` horas en el futuro.
3.  Crearemos una nueva columna `target` (nuestra variable objetivo):
    - `1` si el precio futuro subió (éxito).
    - `0` si el precio futuro bajó o se mantuvo igual (fracaso).

```python
# Este es un bloque de pseudocódigo/guía para el siguiente paso.
# La implementación real dependerá de la cantidad y calidad de los datos.

if not df_analysis.empty:
    # 1. Convertir timestamp a datetime
    df_analysis['timestamp'] = pd.to_datetime(df_analysis['timestamp'])

    # 2. Definir el horizonte de predicción (ej. 4 horas hacia adelante)
    prediction_horizon = 4

    # 3. Crear la columna de precio futuro
    df_analysis['future_price'] = df_analysis['close'].shift(-prediction_horizon)

    # 4. Crear la variable objetivo (target)
    # Éxito si el precio futuro es mayor que el precio de cierre actual
    df_analysis['target'] = (df_analysis['future_price'] > df_analysis['close']).astype(int)

    # 5. Limpiar filas sin target (las últimas 'prediction_horizon' filas)
    df_analysis.dropna(subset=['future_price'], inplace=True)

    print("Variable 'target' creada. Vista previa:")
    display(df_analysis[['timestamp', 'close', 'future_price', 'decision', 'target']].head())
else:
    print("No se puede crear el target porque el dataframe de análisis está vacío.")
```
